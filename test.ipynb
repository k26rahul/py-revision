```python
def constant(n):
  return 1


def logarithmic(n):
  return math.log2(n)


def linear(n):
  return n


def linearithmic(n):
  return n * logarithmic(n)


def quadratic(n):
  return n**2


def cubic(n):
  return n**3


def expo(n):
  return 2 ** n


def factorial(n):
  facto = 1
  for i in range(1, n+1):
    facto *= i
  return facto
```

```python
x = range(1, 6)

_ = plt.plot(x, [constant(n) for n in x])
_ = plt.plot(x, [round(logarithmic(n), 1) for n in x])
_ = plt.plot(x, [linear(n) for n in x])
_ = plt.plot(x, [round(linearithmic(n), 1) for n in x])
_ = plt.plot(x, [quadratic(n) for n in x])
_ = plt.plot(x, [cubic(n) for n in x])
_ = plt.plot(x, [expo(n) for n in x])
_ = plt.plot(x, [factorial(n) for n in x])
```
